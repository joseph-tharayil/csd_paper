# This notebook generates panels g and h of figure 4

```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
def secondDeriv(x):
    delta = x[1]-x[0]
    return np.array([-1,2,-1])/delta**2
```

```python
def fourierTransform(k,delta):
    return 4*(np.sin(k*delta))**2/delta**2
```

```python
def voltageDifferential(radius):
    '''
    This function assumes that a current source is in the plane of one electrode, and the second electrode is offset by d=1
    '''

    v0 = 1/radius
    v1 = 1/np.sqrt(radius**2+1)

    return v0-v1
```

```python
x0 = np.array([-20,0,20])
x1 = np.array([-40,0,40])
x2 = np.array([-80,0,80])
x3 = np.array([-160,0,160])
y0 = secondDeriv(x0)
y1 = secondDeriv(x1)
y2 = secondDeriv(x2)
y3 = secondDeriv(x3)

k = np.linspace(0,0.3)
f0 = fourierTransform(k,20)
f1 = fourierTransform(k,40)
f2 = fourierTransform(k,80)
f3 = fourierTransform(k,160)

fig, ax = plt.subplots(1,2,figsize=(9,3))

markerline, stemlines, baseline = ax[0].stem(x0, y0,basefmt=' ')

# Set colors
plt.setp(markerline, color='C0')      # markers
plt.setp(stemlines, color='C0')      # stems
plt.setp(baseline, color='green')      # baseline

markerline, stemlines, baseline = ax[0].stem(x1, y1,basefmt=' ')

# Set colors
plt.setp(markerline, color='C1')      # markers
plt.setp(stemlines, color='C1')      # stems
plt.setp(baseline, color='green')      # baseline

markerline, stemlines, baseline = ax[0].stem(x2, y2,basefmt=' ')

# Set colors
plt.setp(markerline, color='C2')      # markers
plt.setp(stemlines, color='C2')      # stems
plt.setp(baseline, color='green')      # baseline

markerline, stemlines, baseline = ax[0].stem(x3, y3,basefmt=' ')

# Set colors
plt.setp(markerline, color='C3')      # markers
plt.setp(stemlines, color='C3')      # stems
plt.setp(baseline, color='green')      # baseline

ax[0].set_xlabel('Z')
ax[0].set_ylabel('f(Z)')

ax[0].legend(
    ['d=1','d=2'],
    handlelength=1.0,   # shorter line/marker handle
    handletextpad=0.4,  # space between handle and text
    borderpad=0.3,      # padding inside legend box
    labelspacing=0.3    # vertical space between entries
)


ax[1].plot(k,4*k**2,'k--')
ax[1].plot(k,f0)
ax[1].plot(k,f1)
ax[1].plot(k,f2)
ax[1].plot(k,f3)
ax[1].set_xlabel('Spatial frequency k')
ax[1].set_ylabel('F(k)')
ax[1].set_ylim([0,0.012])

plt.rcParams.update({'font.size': 12})

plt.tight_layout()

plt.savefig('derivativeIllustration.png',dpi=600)
```
