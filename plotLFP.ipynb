```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib

font = {'size'   : 12}

matplotlib.rc('font', **font)

matplotlib.rcParams['savefig.pad_inches'] = 0

%matplotlib inline
```

```python
highRes = np.load('hypothesisTesting/nonshiftedPositions_rerun/total.npy') # Loads LFP and oCSD averaged over trials, 20 um spacing, rho=500um

lfpIdx = np.arange(101) 

totalLFPHighRes = highRes[:,lfpIdx] # LFP, 20 um spacing
totalLFPHighRes -= totalLFPHighRes[0]
```

```python
totalLFPHighRes.shape
```

```python
np.mean(totalLFPHighRes,axis=1).shape
```

```python
plt.figure(figsize=(5,5))
time = np.arange(0,50)
for i in range(101):
    plt.plot(time,totalLFPHighRes[:,i]+i*1e-4,'grey')

plt.ylabel(r'Depth ($\mu$m)')
plt.xlabel('Time (ms)')
ax = plt.gca()
ax.set_yticklabels([])
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
# ax.spines['left'].set_visible(False)
plt.tight_layout()
plt.savefig('hypothesisTesting/nonshiftedPositions_rerun/lfp.png',dpi=600)
```
