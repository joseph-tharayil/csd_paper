```python
import numpy as np
import pandas as pd
```

```python
original = pd.read_csv('electrode_csv_highRes_lfpOnly.csv')
new = pd.read_csv('electrode_csv_highRes_lfpOnly_newAxis.csv')
```

```python
originalCoords = original[['x','y','z']].values
```

```python
newCoords = new[['x','y','z']].values
```

```python
import numpy as np

def angle_between_lines(line1_points, line2_points):
    """
    Calculate the angle between two lines defined by nx3 arrays of points.
    
    Args:
        line1_points: nx3 numpy array of points on line 1
        line2_points: nx3 numpy array of points on line 2
    
    Returns:
        Angle in degrees between the two lines
    """
    # Fit a direction vector to each line using PCA (SVD)
    def fit_direction(points):
        centroid = np.mean(points, axis=0)
        centered = points - centroid
        _, _, Vt = np.linalg.svd(centered)
        return Vt[0]  # First right singular vector = principal direction

    dir1 = fit_direction(line1_points)
    dir2 = fit_direction(line2_points)

    # Clamp dot product to [-1, 1] to avoid floating point issues with arccos
    cos_angle = np.clip(np.dot(dir1, dir2), -1.0, 1.0)

    # Use the acute angle between the lines (lines have no orientation)
    angle_rad = np.arccos(np.abs(cos_angle))
    return np.degrees(angle_rad)


# --- Example usage ---

angle = angle_between_lines(originalCoords, newCoords)
print(f"Angle between lines: {angle:.4f}°")  # → 45.0000°
```
